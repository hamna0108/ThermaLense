```python
# Step 3: Upload and Extract package
from google.colab import files
import zipfile

if not os.path.exists('data/datasets/train.npz'):
    print('Please select and upload heat_fno_seeds_colab_package.zip:')
    uploaded = files.upload()
    for fn in uploaded.keys():
        if fn.endswith('.zip'):
            print(f'Extracting {fn}...')
            with zipfile.ZipFile(fn, 'r') as z:
                z.extractall('.')
            print('Extraction complete!')

!ls -lh data/datasets/
```
